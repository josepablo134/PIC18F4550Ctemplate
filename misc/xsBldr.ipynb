```python
import serial
from pyxsModem import *

def connect():
    return serial.Serial( "COM5", 115200, timeout=1/100_000 )
```

```python
s = connect()
```

```python
if( enter_reprogramming( s, 10 ) ):
    print("reprogramming session active")
else:
    print("timeout reached")
    
```

```python
init_addr = 0x00
end_addr = 0x2000
block_size = 0x40

for block_addr in range( init_addr, end_addr, block_size ):
    if( not write_block( s, block_addr, bytearray(b'Hello world from the UART bootldFEDCBA9876543210FEDCBA9876543210') ) ):
        print(f"error programming block {hex(block_addr)}")
        break
```

```python
mem_dump = bytearray()

init_addr = 0x00
end_addr = 0x2000
block_size = 0x40

for block_addr in range( init_addr, end_addr, block_size ):
    payload = read_block( s , block_addr, 64 )

    if( payload == None ):
        print(f"error reading block {hex(block_addr)}")
        break

    mem_dump += payload

print( mem_dump )
```

```python
s.write( b'A' )
receive_byte( s )
```

```python
byte_arr = bytearray()

single_byte = s.read()
while( single_byte != b'' ):
    byte_arr.append( int.from_bytes( single_byte ) )
    single_byte = s.read()

print_byte_as_hex( byte_arr )
```
